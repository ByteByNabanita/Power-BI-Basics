ANS 1- Power BI is a widely used business intelligence tool by Microsoft that helps convert raw data into interactive reports and visual insights, which drive business decisions. It is one of the most demanding tool for data analysis.


Business use cases of Power BI-

(a) track sales, regions, products, and trends to monitor targets and improve revenue decisions for sales performance dashboard.

(b) analyze employee turnover by department,age, and experience to identify retention issues for HR attrition analysis.

(c) monitor revenue, expenses,and profit trends to support budgeting and financial planning for finance revenue tracking.

ANS 2- The 3 main components of Power BI-

(a) Power BI desktop is a free desktop application where users connect data, clean it using power query and build reports.

(b) Power BI service is a cloud based platform used to publish, share and collaborate on reports created in Power BI desktop.

(c) Power BI mobile is a mobile application  used to view and interact with reports on smartphones and tablets.


ANS 3- Workflow of Power BI-

Connect -> Transform -> Visualize -> Share


ANS 4- Any four data cleaning tasks that can be performed in Power Query-



1.   Removing unnecessary columns
2.   changing data types
3. identifying and replacing null values
4. removing duplicates






ANS 5- Step by step instructions to load the Real_Estate_Dataset :

Step 1: open Power BI. Launch power BI to start a new project

Step 2: click "get data" from home tab.

Step 3: select the data source. Choose the csv file.

Step 4: preview the data, transform data to clean before loading.

Step 5: do the necessary changes and click "close and apply".

ANS 6-  Data view views imported data tables.

Report view creates and design reports.

ANS 8- To Split Owner Name to create two new columns as First Name and Last Name:

from power query editor, select the column -> go to transform tab -> split column -> choose a splitting method: by delimeter -> click on OK and verify the result -> rename the columns -> click close and apply.
